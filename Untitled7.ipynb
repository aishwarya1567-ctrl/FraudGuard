```python
# ================================================================
# PROJECT: Machine Learning-Based Fraud Detection
# DATASET: PaySim1 (simulated mobile-money transactions)
#
# This single cell:
# 1. Downloads the dataset
# 2. Explores and cleans the data
# 3. Creates charts
# 4. Trains Logistic Regression and Random Forest
# 5. Evaluates and compares the models
# 6. Predicts one transaction
# 7. Saves outputs and downloads them as a ZIP
#
# IMPORTANT:
# - PaySim is simulated data, not genuine bank transaction records.
# - This code creates a tabular ML baseline. It does not implement
#   graph traversal or a web application yet.
# - Full PaySim training can take a long time and use substantial RAM.
# ================================================================


# -------------------- STEP 1: INSTALL LIBRARY --------------------
# KaggleHub downloads datasets from Kaggle.
%pip -q install kagglehub


# -------------------- STEP 2: IMPORT LIBRARIES -------------------
import os
import glob
import json
import zipfile
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub
import joblib

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, average_precision_score, classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

# Fixed seed makes random splits and sampling repeatable.
RANDOM_STATE = 42

# Change this to True only if the full dataset is too large or slow
# for your Colab runtime. If True, the model uses a stratified sample.
USE_SAMPLE_FOR_FASTER_DEMO = False
SAMPLE_SIZE = 300_000

# Create a folder where reports, models, and charts will be saved.
OUTPUT_FOLDER = "/content/fraud_detection_outputs"
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

print("=" * 70)
print("PAYSim FRAUD DETECTION PROJECT")
print("=" * 70)


# -------------------- STEP 3: DOWNLOAD DATASET -------------------
print("\nSTEP 1: Downloading PaySim1 dataset...")

# KaggleHub downloads the dataset and returns its folder path.
dataset_path = kagglehub.dataset_download("ealaxi/paysim1")
print("Dataset folder:", dataset_path)

# Search for CSV files instead of assuming a specific filename.
csv_files = glob.glob(os.path.join(dataset_path, "**", "*.csv"), recursive=True)

if len(csv_files) == 0:
    raise FileNotFoundError(
        "No CSV file was found in the downloaded dataset folder. "
        "Check the Kaggle download or upload the CSV manually."
    )

# Select the largest CSV if the downloaded folder contains multiple CSVs.
# The chosen file is printed so this decision is visible.
csv_file_path = max(csv_files, key=os.path.getsize)
print("CSV selected:", csv_file_path)

# Load the CSV into a pandas table called a DataFrame.
transaction_data = pd.read_csv(csv_file_path, low_memory=False)

print("\nDataset downloaded and loaded successfully.")
print("Original dataset shape (rows, columns):", transaction_data.shape)
print("\nFirst five rows:")
display(transaction_data.head())


# -------------------- STEP 4: CHECK REQUIRED COLUMNS -------------
print("\nSTEP 2: Checking required columns...")

# These are the columns used by our baseline model.
feature_columns = [
    "step", "type", "amount", "oldbalanceOrg",
    "newbalanceOrig", "oldbalanceDest", "newbalanceDest"
]
target_column = "isFraud"

required_columns = feature_columns + [target_column]
missing_columns = [
    column for column in required_columns
    if column not in transaction_data.columns
]

if missing_columns:
    raise ValueError(
        "The dataset is missing required columns: " + ", ".join(missing_columns)
    )

print("All required model columns are present.")
print("\nColumn names:")
print(transaction_data.columns.tolist())

print("\nData types:")
print(transaction_data.dtypes)

print("\nBasic numerical statistics:")
display(transaction_data.describe(include="all").transpose())


# -------------------- STEP 5: SAVE ORIGINAL AND CHECK DATA -------
print("\nSTEP 3: Checking data quality...")

# Keep an unchanged copy so we can compare before and after cleaning.
original_data = transaction_data.copy()

# Preserve account identifiers separately for prediction output or future graph work.
identifier_columns = [
    column for column in ["nameOrig", "nameDest"]
    if column in transaction_data.columns
]

print("\nMissing values by column:")
missing_before = transaction_data.isna().sum()
display(missing_before.to_frame("missing_values"))

# Count exact duplicate rows.
duplicates_found = int(transaction_data.duplicated().sum())
print("Exact duplicate rows found:", duplicates_found)

# Check negative amounts and balances, but do not delete rows automatically.
# Unusual values should be investigated rather than silently removed.
for column in [
    "amount", "oldbalanceOrg", "newbalanceOrig",
    "oldbalanceDest", "newbalanceDest"
]:
    if column in transaction_data.columns:
        numeric_values = pd.to_numeric(transaction_data[column], errors="coerce")
        print(f"Negative values in {column}: {int((numeric_values < 0).sum())}")

print("Observed target labels:", sorted(transaction_data["isFraud"].dropna().unique().tolist()))
print("Observed transaction types:", sorted(transaction_data["type"].dropna().astype(str).unique().tolist()))


# -------------------- STEP 6: CLEAN DATA -------------------------
print("\nSTEP 4: Cleaning data...")

rows_before_cleaning = len(transaction_data)

# Remove exact duplicate rows and record how many were removed.
transaction_data = transaction_data.drop_duplicates().copy()
duplicates_removed = rows_before_cleaning - len(transaction_data)

# Convert expected numeric columns to numeric values.
# Unreadable values become NaN so they can be detected and handled.
numeric_columns = [
    "step", "amount", "oldbalanceOrg", "newbalanceOrig",
    "oldbalanceDest", "newbalanceDest", "isFraud"
]
if "isFlaggedFraud" in transaction_data.columns:
    numeric_columns.append("isFlaggedFraud")

for column in numeric_columns:
    transaction_data[column] = pd.to_numeric(
        transaction_data[column], errors="coerce"
    )

# The target must be known and must be either 0 or 1 for supervised learning.
rows_before_target_filter = len(transaction_data)
transaction_data = transaction_data[
    transaction_data["isFraud"].isin([0, 1])
].copy()
invalid_target_rows_removed = rows_before_target_filter - len(transaction_data)
transaction_data["isFraud"] = transaction_data["isFraud"].astype(int)

# Remove rows missing the target or required model inputs.
# We do not remove valid zero balances or large transaction amounts.
rows_before_missing_filter = len(transaction_data)
transaction_data = transaction_data.dropna(
    subset=required_columns
).copy()
missing_required_rows_removed = rows_before_missing_filter - len(transaction_data)

# Keep identifiers aligned with the cleaned rows. They are not model features.
identifier_data = transaction_data[identifier_columns].copy() if identifier_columns else pd.DataFrame(index=transaction_data.index)

cleaning_summary = {
    "rows_before_cleaning": int(rows_before_cleaning),
    "exact_duplicates_found": int(duplicates_found),
    "exact_duplicates_removed": int(duplicates_removed),
    "invalid_target_rows_removed": int(invalid_target_rows_removed),
    "rows_missing_required_values_removed": int(missing_required_rows_removed),
    "rows_after_cleaning": int(len(transaction_data))
}

print("Cleaning summary:")
print(json.dumps(cleaning_summary, indent=2))
print("Shape after cleaning:", transaction_data.shape)
print("Missing values in required columns after cleaning:")
display(transaction_data[required_columns].isna().sum().to_frame("missing_values"))


# -------------------- STEP 7: DATA SUMMARY -----------------------
print("\nSTEP 5: Dataset summary...")

total_transactions = len(transaction_data)
fraud_count = int((transaction_data["isFraud"] == 1).sum())
non_fraud_count = int((transaction_data["isFraud"] == 0).sum())
fraud_percentage = (
    (fraud_count / total_transactions) * 100
    if total_transactions > 0 else 0
)

print("Total transactions:", total_transactions)
print("Fraud transactions:", fraud_count)
print("Non-fraud transactions:", non_fraud_count)
print("Fraud percentage: {:.6f}%".format(fraud_percentage))
print("Transaction types:", sorted(transaction_data["type"].astype(str).unique().tolist()))

# The label imbalance matters: fraud is usually much rarer than non-fraud.
print("\nTarget counts:")
display(
    transaction_data["isFraud"]
    .value_counts()
    .rename(index={0: "Non-fraud (0)", 1: "Fraud (1)"})
    .to_frame("transaction_count")
)


# -------------------- STEP 8: VISUALIZATIONS ---------------------
print("\nSTEP 6: Creating charts...")

# Chart 1: fraud and non-fraud counts.
label_counts = transaction_data["isFraud"].value_counts().reindex([0, 1], fill_value=0)
plt.figure(figsize=(7, 5))
sns.barplot(x=["Non-fraud (0)", "Fraud (1)"], y=label_counts.values)
plt.title("Fraud vs Non-Fraud Transaction Counts")
plt.xlabel("Transaction label")
plt.ylabel("Number of transactions")
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_FOLDER, "fraud_counts.png"), dpi=150)
plt.show()

# Chart 2: transaction type counts.
type_counts = transaction_data["type"].value_counts()
plt.figure(figsize=(9, 5))
sns.barplot(x=type_counts.index, y=type_counts.values)
plt.title("Transaction Type Distribution")
plt.xlabel("Transaction type")
plt.ylabel("Number of transactions")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_FOLDER, "transaction_types.png"), dpi=150)
plt.show()

# Chart 3: transaction amount distribution.
# A log axis helps show the distribution when some transactions are much larger.
plt.figure(figsize=(9, 5))
sns.histplot(data=transaction_data, x="amount", bins=60)
plt.xscale("log")
plt.title("Transaction Amount Distribution (Log Scale)")
plt.xlabel("Transaction amount (log scale)")
plt.ylabel("Number of transactions")
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_FOLDER, "amount_distribution.png"), dpi=150)
plt.show()

# Chart 4: fraud counts by transaction type.
fraud_by_type = (
    transaction_data[transaction_data["isFraud"] == 1]["type"]
    .value_counts()
)
plt.figure(figsize=(9, 5))
sns.barplot(x=fraud_by_type.index, y=fraud_by_type.values)
plt.title("Fraud Counts by Transaction Type")
plt.xlabel("Transaction type")
plt.ylabel("Number of fraud transactions")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_FOLDER, "fraud_by_type.png"), dpi=150)
plt.show()

# Chart 5: numerical correlation heatmap.
# Correlation shows linear association; it does not prove causation.
correlation_columns = [
    "step", "amount", "oldbalanceOrg", "newbalanceOrig",
    "oldbalanceDest", "newbalanceDest", "isFraud"
]
correlation_matrix = transaction_data[correlation_columns].corr(numeric_only=True)
plt.figure(figsize=(10, 7))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Between Numerical Features")
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_FOLDER, "correlation_heatmap.png"), dpi=150)
plt.show()

print("Charts displayed and saved.")


# -------------------- STEP 9: PREPARE ML DATA --------------------
print("\nSTEP 7: Preparing data for machine learning...")

# Features are the inputs used by the model. The target is the answer to predict.
# We exclude isFraud from the inputs because it is the answer.
# We exclude isFlaggedFraud because it is an existing fraud flag that may leak information.
# We exclude nameOrig/nameDest from this baseline because they are raw IDs.
model_data = transaction_data[feature_columns + [target_column]].copy()

# Optional sampling mode for low-memory runtimes. It is OFF by default.
# If enabled, a stratified sample is taken and the report should disclose that.
if USE_SAMPLE_FOR_FASTER_DEMO and len(model_data) > SAMPLE_SIZE:
    sampled_groups = []
    label_proportions = model_data[target_column].value_counts(normalize=True)

    for label, proportion in label_proportions.items():
        rows_for_label = model_data[model_data[target_column] == label]
        sample_count = max(1, int(round(SAMPLE_SIZE * proportion)))
        sample_count = min(sample_count, len(rows_for_label))
        sampled_groups.append(
            rows_for_label.sample(n=sample_count, random_state=RANDOM_STATE)
        )

    model_data = pd.concat(sampled_groups, ignore_index=False)
    print("WARNING: sample mode is ON. Training uses a stratified sample.")
else:
    print("Training will use all cleaned rows.")

X = model_data[feature_columns].copy()
y = model_data[target_column].copy()

# Stratified split keeps approximately the same fraud ratio in train and test.
# 80% is used to train; 20% is kept separate for testing.
# random_state makes the split repeatable.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Fraud percentage in training data: {:.6f}%".format(y_train.mean() * 100))
print("Fraud percentage in testing data: {:.6f}%".format(y_test.mean() * 100))

# Numerical features need numerical preprocessing.
numeric_features = [
    "step", "amount", "oldbalanceOrg",
    "newbalanceOrig", "oldbalanceDest", "newbalanceDest"
]
categorical_features = ["type"]

# The numeric pipeline fills any remaining missing values with the median,
# then scales values to help Logistic Regression.
numeric_preprocessor = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# One-hot encoding turns transaction categories into numerical columns.
# handle_unknown="ignore" avoids an error if an unfamiliar type appears later.
categorical_preprocessor = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

# ColumnTransformer applies the right preprocessing to each type of column.
preprocessor = ColumnTransformer(transformers=[
    ("numeric", numeric_preprocessor, numeric_features),
    ("categorical", categorical_preprocessor, categorical_features)
])


# -------------------- STEP 10: TRAIN MODELS ----------------------
print("\nSTEP 8: Training machine-learning models...")

# Logistic Regression estimates the probability of a transaction being fraud.
# class_weight="balanced" gives extra weight to the less common fraud class.
logistic_model = Pipeline(steps=[
    ("preprocessing", preprocessor),
    ("classifier", LogisticRegression(
        class_weight="balanced",
        max_iter=300,
        random_state=RANDOM_STATE
    ))
])

print("Training Logistic Regression...")
logistic_model.fit(X_train, y_train)
print("Logistic Regression finished.")

# Random Forest combines many decision trees to make predictions.
# The depth limit helps control memory/time for a large dataset.
random_forest_model = Pipeline(steps=[
    ("preprocessing", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=RANDOM_STATE,
        n_jobs=-1
    ))
])

print("Training Random Forest. This may take a while on the full dataset...")
random_forest_model.fit(X_train, y_train)
print("Random Forest finished.")


# -------------------- STEP 11: EVALUATE MODELS -------------------
print("\nSTEP 9: Evaluating models...")

models = {
    "Logistic Regression": logistic_model,
    "Random Forest": random_forest_model
}
metrics_results = {}
predictions_by_model = {}

for model_name, model in models.items():
    print("\nEvaluating:", model_name)

    # Predict class labels and fraud probabilities on the held-out test set.
    predicted_labels = model.predict(X_test)
    predicted_probabilities = model.predict_proba(X_test)[:, 1]

    # Calculate all metrics from actual predictions.
    metrics_results[model_name] = {
        "accuracy": accuracy_score(y_test, predicted_labels),
        "precision": precision_score(y_test, predicted_labels, zero_division=0),
        "recall": recall_score(y_test, predicted_labels, zero_division=0),
        "f1_score": f1_score(y_test, predicted_labels, zero_division=0),
        "roc_auc": roc_auc_score(y_test, predicted_probabilities),
        "average_precision": average_precision_score(y_test, predicted_probabilities)
    }

    predictions_by_model[model_name] = {
        "labels": predicted_labels,
        "probabilities": predicted_probabilities
    }

    for metric_name, metric_value in metrics_results[model_name].items():
        print(f"{metric_name}: {metric_value:.6f}")

    print("\nClassification report:")
    print(classification_report(
        y_test,
        predicted_labels,
        labels=[0, 1],
        target_names=["Non-fraud", "Fraud"],
        zero_division=0
    ))

    # Confusion matrix: rows are actual labels; columns are predicted labels.
    # False negative means a fraud transaction was missed.
    cm = confusion_matrix(y_test, predicted_labels, labels=[0, 1])
    print("Confusion matrix:")
    display(pd.DataFrame(
        cm,
        index=["Actual non-fraud", "Actual fraud"],
        columns=["Predicted non-fraud", "Predicted fraud"]
    ))

    plt.figure(figsize=(6, 5))
    sns.heatmap(
        cm, annot=True, fmt="d", cmap="Blues",
        xticklabels=["Predicted non-fraud", "Predicted fraud"],
        yticklabels=["Actual non-fraud", "Actual fraud"]
    )
    plt.title("Confusion Matrix - " + model_name)
    plt.xlabel("Predicted label")
    plt.ylabel("Actual label")
    plt.tight_layout()
    safe_name = model_name.lower().replace(" ", "_")
    plt.savefig(os.path.join(OUTPUT_FOLDER, safe_name + "_confusion_matrix.png"), dpi=150)
    plt.show()

# Compare the two models in one table.
metrics_table = pd.DataFrame(metrics_results).transpose()
metrics_table.index.name = "Model"
print("\nMODEL COMPARISON")
display(metrics_table.sort_values("average_precision", ascending=False))


# -------------------- STEP 12: SELECT FINAL MODEL ----------------
print("\nSTEP 10: Selecting the final baseline model...")

# Average precision is used as the primary comparison metric because fraud is rare.
# This is a simple baseline selection rule; a production project should use validation data.
final_model_name = max(
    metrics_results,
    key=lambda name: metrics_results[name]["average_precision"]
)
final_model = models[final_model_name]

# Keep the default threshold fixed at 0.50; it is not tuned using the test set.
decision_threshold = 0.50
final_probabilities = final_model.predict_proba(X_test)[:, 1]
final_predictions = (final_probabilities >= decision_threshold).astype(int)

final_metrics = {
    "accuracy": float(accuracy_score(y_test, final_predictions)),
    "precision": float(precision_score(y_test, final_predictions, zero_division=0)),
    "recall": float(recall_score(y_test, final_predictions, zero_division=0)),
    "f1_score": float(f1_score(y_test, final_predictions, zero_division=0)),
    "roc_auc": float(roc_auc_score(y_test, final_probabilities)),
    "average_precision": float(average_precision_score(y_test, final_probabilities))
}

print("Selected model:", final_model_name)
print("Selection criterion: highest average precision.")
print("Decision threshold:", decision_threshold)
print("Final metrics:")
for metric_name, metric_value in final_metrics.items():
    print(f"{metric_name}: {metric_value:.6f}")


# -------------------- STEP 13: FEATURE IMPORTANCE ----------------
print("\nSTEP 11: Random Forest feature importance...")

# Feature importance shows how much the fitted Random Forest used each encoded feature.
# It does not prove that a feature causes fraud.
rf_classifier = random_forest_model.named_steps["classifier"]
rf_preprocessor = random_forest_model.named_steps["preprocessing"]
encoded_feature_names = rf_preprocessor.get_feature_names_out()

feature_importance_table = pd.DataFrame({
    "feature": encoded_feature_names,
    "importance": rf_classifier.feature_importances_
}).sort_values("importance", ascending=False)

print("Top 20 features:")
display(feature_importance_table.head(20))

top_features = feature_importance_table.head(20).sort_values("importance")
plt.figure(figsize=(10, 7))
plt.barh(top_features["feature"], top_features["importance"])
plt.title("Top 20 Random Forest Feature Importances")
plt.xlabel("Importance")
plt.ylabel("Encoded feature")
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_FOLDER, "random_forest_feature_importance.png"), dpi=150)
plt.show()


# -------------------- STEP 14: PREDICT ONE TRANSACTION ------------
print("\nSTEP 12: Example transaction prediction...")

# Choose one held-out test transaction. Its actual label is kept separate.
example_features = X_test.iloc[[0]].copy()
example_actual_label = int(y_test.iloc[0])

# The model receives only input features, never the actual answer.
example_probability = float(final_model.predict_proba(example_features)[0, 1])
example_predicted_label = int(example_probability >= decision_threshold)

print("Input transaction features:")
display(example_features)
print("Actual fraud label:", example_actual_label)
print("Predicted label:", example_predicted_label)
print("Predicted fraud probability: {:.6f}".format(example_probability))
print("Was the prediction correct?", example_actual_label == example_predicted_label)

# Reusable function: give it a dictionary with all seven feature values.
def predict_transaction(transaction_dictionary):
    # Convert one transaction dictionary to a one-row table.
    one_row = pd.DataFrame([transaction_dictionary], columns=feature_columns)

    # Predict the probability of fraud and apply the same saved threshold.
    fraud_probability = float(final_model.predict_proba(one_row)[0, 1])
    predicted_class = int(fraud_probability >= decision_threshold)

    return {
        "predicted_class": predicted_class,
        "predicted_fraud_probability": fraud_probability,
        "decision_threshold": decision_threshold
    }

print("\nExample using the prediction function:")
print(predict_transaction(example_features.iloc[0].to_dict()))


# -------------------- STEP 15: SAVE OUTPUTS -----------------------
print("\nSTEP 13: Saving model, reports, metrics, and predictions...")

# Save the complete pipeline so future code can load preprocessing and model together.
model_path = os.path.join(OUTPUT_FOLDER, "final_fraud_detection_model.joblib")
joblib.dump(final_model, model_path)

# Save model comparison and final metrics.
metrics_table.to_csv(os.path.join(OUTPUT_FOLDER, "model_comparison_metrics.csv"))
with open(os.path.join(OUTPUT_FOLDER, "final_model_metrics.json"), "w", encoding="utf-8") as file:
    json.dump(final_metrics, file, indent=2)

# Save the final classification report and confusion matrix.
report_text = classification_report(
    y_test, final_predictions,
    labels=[0, 1],
    target_names=["Non-fraud", "Fraud"],
    zero_division=0
)
with open(os.path.join(OUTPUT_FOLDER, "final_classification_report.txt"), "w", encoding="utf-8") as file:
    file.write(report_text)

final_cm = confusion_matrix(y_test, final_predictions, labels=[0, 1])
pd.DataFrame(
    final_cm,
    index=["Actual non-fraud", "Actual fraud"],
    columns=["Predicted non-fraud", "Predicted fraud"]
).to_csv(os.path.join(OUTPUT_FOLDER, "final_confusion_matrix.csv"))

# Save feature importance and cleaning information.
feature_importance_table.to_csv(
    os.path.join(OUTPUT_FOLDER, "random_forest_feature_importance.csv"),
    index=False
)
with open(os.path.join(OUTPUT_FOLDER, "cleaning_summary.json"), "w", encoding="utf-8") as file:
    json.dump(cleaning_summary, file, indent=2)

# Save configuration so the backend can use the same columns and threshold.
configuration = {
    "selected_model": final_model_name,
    "feature_columns": feature_columns,
    "target_column": target_column,
    "decision_threshold": decision_threshold,
    "random_state": RANDOM_STATE,
    "sample_mode_enabled": USE_SAMPLE_FOR_FASTER_DEMO,
    "training_rows": int(len(X_train)),
    "testing_rows": int(len(X_test)),
    "dataset_note": "PaySim is simulated transaction data."
}
with open(os.path.join(OUTPUT_FOLDER, "model_configuration.json"), "w", encoding="utf-8") as file:
    json.dump(configuration, file, indent=2)

# Save test predictions and probabilities.
test_prediction_table = X_test.copy()
test_prediction_table["actual_isFraud"] = y_test
test_prediction_table["predicted_isFraud"] = final_predictions
test_prediction_table["predicted_fraud_probability"] = final_probabilities

# Reattach identifiers using original row indexes when they are available.
if identifier_columns:
    ids_for_test = identifier_data.loc[X_test.index].copy()
    ids_for_test.index = test_prediction_table.index
    test_prediction_table = pd.concat([ids_for_test, test_prediction_table], axis=1)

test_prediction_table.to_csv(
    os.path.join(OUTPUT_FOLDER, "test_set_predictions.csv"),
    index=False
)

# Save a readable text summary with actual computed metrics.
summary_lines = [
    "PAYSim Fraud Detection - Results Summary",
    "========================================",
    "Dataset: PaySim1 simulated mobile-money transactions",
    f"Rows before cleaning: {len(original_data)}",
    f"Rows after cleaning: {len(transaction_data)}",
    f"Fraud records: {fraud_count}",
    f"Non-fraud records: {non_fraud_count}",
    f"Fraud percentage: {fraud_percentage:.6f}%",
    f"Duplicate rows removed: {duplicates_removed}",
    f"Invalid target rows removed: {invalid_target_rows_removed}",
    f"Rows missing required values removed: {missing_required_rows_removed}",
    "Features: " + ", ".join(feature_columns),
    f"Training rows: {len(X_train)}",
    f"Testing rows: {len(X_test)}",
    f"Selected model: {final_model_name}",
    f"Decision threshold: {decision_threshold}",
    "",
    "Final metrics:"
]
for metric_name, metric_value in final_metrics.items():
    summary_lines.append(f"{metric_name}: {metric_value:.6f}")

summary_lines += [
    "",
    "Limitations:",
    "- PaySim is simulated data, not genuine bank transactions.",
    "- This notebook builds a tabular ML baseline, not graph traversal.",
    "- Raw account IDs were not used as baseline model features.",
    "- Results should not be treated as production fraud-detection performance."
]

with open(os.path.join(OUTPUT_FOLDER, "project_results_report.txt"), "w", encoding="utf-8") as file:
    file.write("\n".join(summary_lines))

print("Saved output files:")
for file_name in sorted(os.listdir(OUTPUT_FOLDER)):
    print(" -", file_name)


# -------------------- STEP 16: ZIP AND DOWNLOAD ------------------
print("\nSTEP 14: Creating ZIP file for download...")

zip_path = "/content/fraud_detection_outputs.zip"
with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zip_file:
    for folder_name, subfolders, file_names in os.walk(OUTPUT_FOLDER):
        for file_name in file_names:
            full_path = os.path.join(folder_name, file_name)
            relative_path = os.path.relpath(full_path, OUTPUT_FOLDER)
            zip_file.write(full_path, arcname=relative_path)

print("ZIP created:", zip_path)
print("ZIP size (MB):", round(os.path.getsize(zip_path) / (1024 * 1024), 2))

# Google Colab will start downloading the ZIP to your computer.
from google.colab import files
files.download(zip_path)


# -------------------- STEP 17: FINAL SUMMARY ---------------------
print("\n" + "=" * 70)
print("FINAL PROJECT SUMMARY")
print("=" * 70)
print("Rows before cleaning:", len(original_data))
print("Rows after cleaning:", len(transaction_data))
print("Fraud transactions:", fraud_count)
print("Non-fraud transactions:", non_fraud_count)
print("Fraud percentage: {:.6f}%".format(fraud_percentage))
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
print("Models trained: Logistic Regression and Random Forest")
print("Selected final model:", final_model_name)
print("Decision threshold:", decision_threshold)

print("\nActual model comparison metrics:")
display(metrics_table)

print("\nFinal model test metrics:")
for metric_name, metric_value in final_metrics.items():
    print(f"{metric_name}: {metric_value:.6f}")

print("\nOutput folder:", OUTPUT_FOLDER)
print("ZIP file:", zip_path)
print("\nIMPORTANT: This is a baseline model using simulated PaySim data.")
print("Graph traversal and the web application are separate future stages.")
```
